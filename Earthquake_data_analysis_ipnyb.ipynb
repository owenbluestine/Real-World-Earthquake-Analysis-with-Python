```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV // Bring over data
earthquakes_raw = pd.read_csv("EQDATA.csv")

print("Raw Dataset Shape:", earthquakes_raw.shape)
display(earthquakes_raw.head())

# Filter data
data_targets = [
    "time",
    "latitude",
    "longitude",
    "depth",
    "mag",
    "place",
    "type"
]

# Creating usable copy // Inspecting Data
earthquakes = earthquakes_raw[data_targets].copy()
print("Dataset Shape:", earthquakes.shape)
print("\nData types:")
print(earthquakes.dtypes)

print("\nMissing Values:")
print(earthquakes.isna().sum())

print("\nDuplicate Rows:", earthquakes.duplicated().sum())

display(earthquakes.describe())
display(earthquakes.head())

earthquake_count = len(earthquakes)

# Calculating Mag mean/min/max, Depth avg/min/max
mag_avg = earthquakes["mag"].mean()
mag_max = earthquakes["mag"].max()
mag_min = earthquakes["mag"].min()
mag_mode = earthquakes["mag"].mode()

depth_avg = earthquakes["depth"].mean()
depth_max = earthquakes["depth"].max()
depth_min = earthquakes["depth"].min()

print("Total Earthquakes :", earthquake_count)

print(f"\nAverage Magnitude: {mag_avg:.2f}")
print(f"\nMaximum Magnitude: {mag_max:.2f}")
print(f"\nMinimum Magnitude: {mag_min:.2f}")
display(mag_mode)

print(f"\nAverage Depth: {depth_avg:.2f}km")
print(f"\nMaximum Depth: {depth_max:.2f}km")
print(f"\nMinimum Depth: {depth_min:.2f}km")

# Magnitude Classification
def magnitude_classification(m):
  if pd.isna(m):
    return "Error"
  elif m < 3:
    return "Micro"
  elif m >= 3 and m < 4:
    return "Minor"
  elif m >= 4 and m < 5.5:
    return "Moderate"
  elif m >= 5.5 and m < 6:
    return "Strong"
  else:
    return "Massive"

earthquakes["mag_class"] = earthquakes["mag"].apply(magnitude_classification)

# Depth Classification
def depth_classification(d):
  if pd.isna(d):
    return "Error"
  elif d <= 70:
    return "Shallow"
  elif d > 70 and d <= 300:
    return "Intermediate"
  else:
    return "Deep"

earthquakes["depth_class"] = earthquakes["depth"].apply(depth_classification)

# Organizing Magnitude Classifications
mag_classes= [
    "Micro",
    "Minor",
    "Moderate",
    "Strong",
    "Massive",
    "Error"
]
mag_classes_limits = [2.5,3,3.5,4,4.5,5,5.5,6]
mag_counts = []

for mag_type in mag_classes:
  count = (earthquakes["mag_class"] == mag_type).sum()
  mag_counts.append(count)
  print(f"{mag_type}: {count} earthquakes ")

# Organizing Depth Classifications
depth_classes= [
    "Shallow",
    "Intermediate",
    "Deep",
    "Error"
]

depth_counts = []

for depth_type in depth_classes:
  d_count = (earthquakes["depth_class"] == depth_type).sum()
  depth_counts.append(d_count)
  print(f"{depth_type}: {d_count} earthquakes")

# Filtering Mag/Depth classifications for high risk earthquakes
def engineering_filter (data):
  target_quakes = data[
      (data["mag"] >= 4) & (data["depth"] < 50)
      ]
  return target_quakes

engineering_quakes = engineering_filter(earthquakes)
print("Earthquakes with Mag >= 4 and Depth < 50km:")
print("Matching earthquakes:", len(engineering_quakes))

display(engineering_quakes)

# Displaying largest magnitude and depth in dataset
largest_quakes = earthquakes [
    earthquakes["mag"] == earthquakes["mag"].max()
]

display(largest_quakes)

deepest_quakes = earthquakes [
    earthquakes["depth"] == earthquakes["depth"].max()
]

display(deepest_quakes)

# Histographing Magnitudes
plt.figure(figsize=(9,5))
plt.hist(earthquakes["mag"], bins = 20 , edgecolor = "black")
plt.xlabel("Magnitude")
plt.ylabel("Number of Earthquakes")
plt.title("Magnitude vs Frequency Mag: 2.5+ (Sept 2026)")

# Scatter plotting Mag vs Depth
plt.figure(figsize=(10,3))
plt.scatter(earthquakes["depth"],earthquakes["mag"], s = 4)
plt.ylabel("Magnitude")
plt.xlabel("Depth (km)")
plt.title("Magnitude vs Depth (Sept 2026)")

# Bar Graph of Magnitude Classifications
plt.figure(figsize=(6,4))
plt.bar(mag_classes, mag_counts)
plt.xlabel("Magnitude Classification")
plt.ylabel("Number of Earthquakes")
plt.title("Magnitude Classifications (Sept 2026)")

# Bar Graphing of Depth Classifications
plt.figure(figsize=(6,4))
plt.bar(depth_classes,depth_counts)
plt.xlabel("Depth Classification")
plt.ylabel("Number of Earthquakes")
plt.title("Depth Classifications of Earthquakes (Sept 2026)")

display(earthquakes["mag"].max())

```
